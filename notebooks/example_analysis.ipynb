# Accretion Model Grid — Example Analysis

This notebook demonstrates the basic workflow for working with the STARS
accretion model grid (Dimoff et al. 2025):

1. **Load** a single model file
2. **Extract** surface parameters across the grid
3. **Extract** surface chemical abundances
4. **Plot** an HR diagram

Before running this notebook, make sure you have:
- Installed the required packages: `pip install -r requirements.txt`
- Downloaded the model grid into `data/models/` (see `data/README.md`)

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

# Make sure the repository root is on the Python path
repo_root = Path().resolve().parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from scripts.read_model import load_model
from scripts.extract_surface_params import extract_surface_params
from scripts.extract_surface_abundances import extract_surface_abundances
from scripts.plot_hr_diagram import plot_hr_diagram

DATA_DIR = repo_root / 'data' / 'models'
```

## 1. Load a Single Model

Use `load_model` to read one model output file into a pandas DataFrame.

```python
# Adjust the path to point to an actual model file in your downloaded grid
example_file = DATA_DIR / 'Z0.014' / 'M01.0' / 'mdot1e-7' / 'model.dat'

if example_file.exists():
    df = load_model(example_file)
    print(f'Loaded {len(df)} time-steps')
    display(df.head())
else:
    print(f'File not found: {example_file}')
    print('Please download the model grid — see data/README.md')
```

## 2. Extract Surface Parameters Across the Grid

```python
if DATA_DIR.is_dir():
    params_df = extract_surface_params(DATA_DIR, all_timesteps=False)
    print(f'Extracted {len(params_df)} rows')
    display(params_df.head())
else:
    print(f'Data directory not found: {DATA_DIR}')
```

## 3. Extract Surface Abundances

```python
if DATA_DIR.is_dir():
    abund_df = extract_surface_abundances(DATA_DIR, all_timesteps=False)
    print(f'Extracted {len(abund_df)} rows')
    display(abund_df.head())
else:
    print(f'Data directory not found: {DATA_DIR}')
```

## 4. Plot the HR Diagram

Plot evolutionary tracks for solar-metallicity models.

```python
if DATA_DIR.is_dir():
    fig = plot_hr_diagram(
        data_dir=DATA_DIR,
        metallicity=0.014,
        output=None,   # don't save — just display inline
        show=False,
    )
    plt.show()
else:
    print(f'Data directory not found: {DATA_DIR}')
```
